```python
#kaushik tamgadge
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("SeoulBikeData.csv", encoding="ISO-8859-1")
```

```python
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)
```

```python
# Histogram for Rented Bike Count
sns.histplot(df["Rented Bike Count"], kde=True, bins=30, color="blue")
plt.title("Distribution of Rented Bike Count")
plt.show()
```

```python
# Scatterplot: Temperature vs Bike Count
sns.scatterplot(x="Temperature(°C)", y="Rented Bike Count", data=df, hue="Seasons", alpha=0.6)
plt.title("Temperature vs Rented Bike Count")
plt.show()
```

```python
# Boxplot: Seasons vs Bike Count
sns.boxplot(x="Seasons", y="Rented Bike Count", data=df, palette="muted")
plt.title("Rented Bike Count Across Seasons")
plt.show()
```

```python
# Lineplot: Hour vs Average Bike Count
sns.lineplot(x="Hour", y="Rented Bike Count", data=df, estimator="mean", ci=None, marker="o")
plt.title("Average Rented Bike Count by Hour")
plt.show()
```

```python
# Heatmap for correlation
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
